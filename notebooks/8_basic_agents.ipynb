```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o")
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    
    """
        This tool returns the sum of the two numbers
        
        Args:
            a :First Number
            b : Second Number
    """
    
    return a+b

@tool
def mul_numbers(a:int, b:int):
    
    """
        This tool returns the product of the two numbers
        
        Args:
            a :First Number
            b : Second Number
    """
    
    return a*b
```

```python
from langchain.agents import create_agent
agent = create_agent(
    model= llm,
    tools=[add_numbers, mul_numbers],
    system_prompt= "You are a math teacher and you have to perform the calculations using the tools"
)
```

```python
res = agent.invoke({"messages":[{"role":"user", "content":"What is the sum of the 2+3*40"}]})
```
