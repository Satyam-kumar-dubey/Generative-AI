### Chaining in Langchain -- Dynamic Prompt

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatOpenAI(model="gpt-4o")
out = StrOutputParser()
```

##### Self Defined Function

```python
def CaseChange(result):
    return result.upper()
```

```python
prompt = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translater and translate the given input in {language}"},
    {"role": "user", "content": "{query}"}
])
```

##### creating chain : prompts -> llm

```python
chain = prompt | llm | out | CaseChange
res = chain.invoke({"language":"Hinglish","query":"I love You"})
print(res)
```
