##### Load env Variables

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4.1")
```

```python
res = llm.invoke("Who is Virat Kohli?")
print(res.content)
```

#### Google Gemini Model

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
res = llm.invoke("What is an array ?")
print(res.content)
```

#### Static Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
geminiLLM = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
prompts = [
    ("system", "You are a senior C++ Developer"),
    ("user", "How to sort an Array ?")
]
res = geminiLLM.invoke(prompts)
print(res.content)
```
